## Model

$n$ identical herders share a pasture. Each herder $i$ chooses $a_i \geq 0$ animals.  
Total animals: $A = \sum_{j=1}^{n} a_j$.  
Value per animal falls with congestion: $v(A) = 10 - A$.  
Cost per animal: $c = 2$ (constant).

Each herder's profit: $\pi_i = a_i\,[v(A) - c]$.

**Rent** $= A\,[v(A) - c]$: the total value the pasture generates above costs.

### Three regimes

| Regime | Decision rule | FOC | Solution | Rent |
|---|---|---|---|---|
| **Social planner** | max total rent | $v(A) + A\,v'(A) = c$ | $A^* = 4$ | $16$ |
| **Nash equilibrium** ($n$ herders) | each max own $\pi_i$ | $v(A) + \tfrac{A}{n}\,v'(A) = c$ | $A^N = \dfrac{8n}{n+1}$ | $\dfrac{64n}{(n+1)^2}$ |
| **Open access** ($n \to \infty$) | entry until $\pi = 0$ | $v(A) = c$ | $A_{OA} = 8$ | $0$ |

### Why the Nash equilibrium is inefficient

The herder's FOC has $\tfrac{A}{n}\,v'$ where the planner's has $A\,v'$. Since $v' < 0$, the herder **subtracts less** — she internalises only $1/n$ of the congestion externality. Result: $A^* < A^N < A_{OA}$.

- **Private benefit** of one more animal: $+1$ (full value captured)
- **Private cost**: only $1/n$ of the damage
- **Social cost**: the full damage $(-1)$ — but the herder ignores the $(n-1)/n$ that falls on others

The tragedy is not that the pasture is necessarily destroyed. It is that the **rent** is progressively **wasted**, because each herder's private cost understates the true social cost.

```python
# Install ipywidgets for JupyterLite (no-op in regular Jupyter)
try:
    import piplite
    await piplite.install(['ipywidgets'])
except ImportError:
    pass
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# UQ colours
UQ_PURPLE = "#512478"
UQ_CYAN   = "#0099CC"

# Model: v(A) = 10 - A, cost c = 2
V_MAX = 10
C = 2
A_STAR = (V_MAX - C) / 2          # = 4
A_OA   = V_MAX - C                # = 8
RENT_STAR = A_STAR * (V_MAX - A_STAR - C)  # = 16

def nash_A(n):
    return (V_MAX - C) * n / (n + 1)

def nash_rent(n):
    return (V_MAX - C)**2 * n / (n + 1)**2
```

## 1. Summary table

How the Nash equilibrium changes as the number of herders $n$ grows:

```python
print(f"{'n':>5s}  {'A^N':>6s}  {'Rent':>7s}  {'% lost':>7s}")
print("-" * 30)
for n in [1, 2, 3, 5, 10, 20, 50, 100]:
    An = nash_A(n)
    Rn = nash_rent(n)
    pct = (1 - Rn / RENT_STAR) * 100
    print(f"{n:5d}  {An:6.2f}  {Rn:7.2f}  {pct:6.1f}%")
print(f"{'∞':>5s}  {A_OA:6.2f}  {'0.00':>7s}  {'100.0%':>7s}")
```

## 2. Interactive simulation

Use the slider to change the number of herders and watch:
- **Left:** total animals converging toward open access as $n$ grows
- **Centre:** rent collapsing toward zero
- **Right:** the rent curve — where the Nash equilibrium sits between the planner's optimum and open access

```python
def simulate(n=5):
    An = nash_A(n)
    Rn = nash_rent(n)
    pct_lost = (1 - Rn / RENT_STAR) * 100

    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 4.5))
    fig.suptitle(f"What happens with n = {n} herder{'s' if n > 1 else ''}?",
                 fontsize=14, fontweight="bold", color=UQ_PURPLE)

    # ── Left: total animals vs n ──
    n_grid = np.arange(1, 101)
    ax1.fill_between(n_grid, nash_A(n_grid), A_STAR,
                     alpha=0.15, color=UQ_PURPLE)
    ax1.plot(n_grid, nash_A(n_grid), color=UQ_PURPLE, lw=2.5)
    ax1.axhline(A_STAR, color="grey", ls="--", lw=1)
    ax1.axhline(A_OA, color=UQ_CYAN, ls="--", lw=1)
    ax1.text(103, A_STAR, "Efficient $A^*=4$", va="center", fontsize=9, color="grey")
    ax1.text(103, A_OA, "Open access $A_{OA}=8$", va="center", fontsize=9, color=UQ_CYAN)
    ax1.plot(n, An, "o", color=UQ_PURPLE, ms=12, zorder=5)
    ax1.annotate(f"  $A^N = {An:.1f}$", xy=(n, An), fontsize=11,
                 fontweight="bold", color=UQ_PURPLE, va="bottom")
    ax1.set_xlabel("Number of herders $n$", fontsize=11)
    ax1.set_ylabel("Total animals on the pasture", fontsize=11)
    ax1.set_xlim(1, 100)
    ax1.set_ylim(2, 9)
    ax1.set_title("More herders → more animals", color=UQ_PURPLE)

    # ── Centre: rent vs n ──
    rent_grid = nash_rent(n_grid)
    ax2.fill_between(n_grid, rent_grid, RENT_STAR,
                     alpha=0.15, color="#CC3333")
    ax2.plot(n_grid, rent_grid, color=UQ_CYAN, lw=2.5)
    ax2.axhline(RENT_STAR, color="grey", ls="--", lw=1)
    ax2.text(103, RENT_STAR, f"Max rent = {RENT_STAR:.0f}", va="center", fontsize=9, color="grey")
    ax2.axhline(0, color="grey", lw=0.5)
    ax2.plot(n, Rn, "o", color=UQ_CYAN, ms=12, zorder=5)
    ax2.annotate(f"  Rent = {Rn:.1f}  ({pct_lost:.0f}% lost)",
                 xy=(n, Rn), fontsize=11, fontweight="bold",
                 color=UQ_CYAN, va="top")
    ax2.set_xlabel("Number of herders $n$", fontsize=11)
    ax2.set_ylabel("Rent ($)", fontsize=11)
    ax2.set_xlim(1, 100)
    ax2.set_ylim(-1, 18)
    ax2.set_title("More herders → less rent", color=UQ_PURPLE)

    # ── Right: rent curve with A*, A^N, A_OA ──
    A_grid = np.linspace(0, 10, 300)
    rent_curve = A_grid * (V_MAX - A_grid - C)
    ax3.plot(A_grid, rent_curve, color=UQ_PURPLE, lw=2.5)
    ax3.axhline(0, color="grey", lw=0.5)

    # Shade rent wasted between A* and A^N
    mask = (A_grid >= A_STAR) & (A_grid <= An)
    ax3.fill_between(A_grid[mask], rent_curve[mask], 0,
                     alpha=0.20, color="#CC3333")

    # Planner
    ax3.plot(A_STAR, RENT_STAR, "s", color="grey", ms=11, zorder=5)
    ax3.annotate(f"  Planner $A^*={A_STAR:.0f}$\n  Rent = {RENT_STAR:.0f}",
                 xy=(A_STAR, RENT_STAR), fontsize=9, color="grey", va="bottom")

    # Nash
    ax3.plot(An, Rn, "o", color=UQ_CYAN, ms=11, zorder=5)
    ax3.annotate(f"  Nash $A^N={An:.1f}$\n  Rent = {Rn:.1f}",
                 xy=(An, Rn), fontsize=9, color=UQ_CYAN, va="top")

    # Open access
    ax3.plot(A_OA, 0, "D", color="#CC3333", ms=9, zorder=5)
    ax3.annotate(f"  Open access\n  $A_{{OA}}={A_OA:.0f}$, Rent = 0",
                 xy=(A_OA, 0), fontsize=9, color="#CC3333", va="bottom")

    ax3.set_xlabel("Total animals $A$", fontsize=11)
    ax3.set_ylabel("Rent ($)", fontsize=11)
    ax3.set_xlim(0, 10)
    ax3.set_ylim(-6, 20)
    ax3.set_title("Planner vs Nash vs Open Access", color=UQ_PURPLE)

    plt.tight_layout()
    plt.show()

interact(simulate, n=IntSlider(min=1, max=100, step=1, value=5,
                               description="Herders n:",
                               style={"description_width": "80px"},
                               layout={"width": "400px"}));
```

## 3. Key takeaways

1. **The tragedy is continuous, not binary.** With just 5 herders, 44% of the rent is already gone.
2. **The wedge grows with $n$.** Each herder internalises only $1/n$ of the congestion externality.
3. **Nash → open access as $n \to \infty$.** The $1/n$ term vanishes and rent is fully dissipated.
4. **The rent curve shows the full picture.** The Nash point slides from the planner's peak ($A^*$) toward open access ($A_{OA}$) as $n$ grows. The red shading shows exactly how much rent is wasted.
5. **Sole owner = planner.** When $n = 1$, private cost equals social cost — no tragedy.
